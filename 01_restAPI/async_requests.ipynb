# Async HTTP Requests
Agents often need to call multiple APIs at the same time.
Async lets you do that without waiting one by one.

# Basic Async GET

```python
import httpx
import asyncio

async def fetch_post(post_id):
    async with httpx.AsyncClient() as client:
        response = await client.get(
            f"https://jsonplaceholder.typicode.com/posts/{post_id}"
        )
        return response.json()

result = await fetch_post(1)
print(result)
```

# Parallel Requests

```python
async def fetch_multiple_posts(post_ids):
    async with httpx.AsyncClient() as client:
        tasks = [
            client.get(f"https://jsonplaceholder.typicode.com/posts/{id}")
            for id in post_ids
        ]
        responses = await asyncio.gather(*tasks)
        return [r.json() for r in responses]
    
results = await fetch_multiple_posts([1,2,3,4,5])
print(f"Fetched {len(results)} posts simultaneously")
for post in results:
    print(f"  - {post['title'][:40]}")
```

## Why does this matter for agents?
When your agent needs to search 5 sources at once,
async cuts the wait time from 5x to ~1x.

Sequential = slow. Parallel = fast.
