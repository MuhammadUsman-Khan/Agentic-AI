# GET Request

````python
import requests

response = requests.get("https://jsonplaceholder.typicode.com/posts/1")

print(response.status_code)
print(response.json())
````

## POST Request
Sending data to an API — this is how you'll send prompts to LLM APIs later.

````python
payload = {
    "title": "My Agent",
    "body": "This is an Agentic AI Post",
    "userId": 1
}

response = requests.post(
    "https://jsonplaceholder.typicode.com/posts",
    json=payload
)

print("Status:", response.status_code)
print("Created:", response.json())
````


## Headers
Most real APIs (including Groq) require headers — especially Authorization.

```
headers = {
    "Authorization": "Bearer YOUR_API_KEY",
    "Content-Type": "application/json"
}
```

````python
response = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    params={"UserId": 1}
)

print("Posts by user 1:", len(response.json()))
````
