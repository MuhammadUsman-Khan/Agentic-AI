# Tool Schemas

Before the LLM can use a tool, it needs to know:
1. What is the tool called?
2. What does it do?
3. What arguments does it need?
4. Which arguments are required?

You define all of this as a JSON schema.
The LLM reads it and decides when and how to use it.



# Anatomy of a tool schema

```python
from traitlets.utils import descriptions

calculator_tool = {
    "type": "function",
    "function": {
        "name": "calculate", # must match with Python function name
        "description": "Performs basic math operations. Use this whenever the user asks to calculate, compute or solve a math problem.", # be specific - LLM reads this to decide when to use it
        "parameters": {
            "type": "object",
            "properties": {
                "operation": {
                    "type": "string",
                    "enum": ["add", "subtract", "multiply", 'divide'], # restrict the choices
                    "description": "The math operation to perform"
                },
                "a": {
                    "type": "integer",
                    "description": "First Number"
                },
                "b": {
                    "type": "integer",
                    "description": "Second Number"
                }
            },
            "required": ["operation", "a", "b"] # LLM must provide these
        }
    }
}

print("Tool schema defined!")
print(f"Tool name: {calculator_tool['function']["name"]}")
print(f"Required args: {calculator_tool['function']['parameters']["required"]}")
```

# Weather tool schema

```python
weather_tool = {
    "type": "function",
    "function": {
        "name": "get_weather",
        "description": "Gets current weather for a city. Use this when user asks about weather, temperature or climate in any location",
        "parameters": {
            "type": "object",
            "properties": {
                "city": {
                    "type": "string",
                    "description": "Name of the city e.g. Karachi, London, New York"
                },
                "unit": {
                    "type": "string",
                    "enum": ["celcius", "fahrenheit"],
                    "description": "Temperature unit"
                }
            },
            "required": ["city"] # unit is optional
        }
    }
}
```

# Search tool schema

```python
search_tool = {
    "type": "function",
    "function": {
        "name": "search_web",
        "description": "Searches the web for current information. Use this when user asks about recent events, news, or anything you don't know",
        "parameters": {
            "type": "object",
            "properties": {
                "query": {
                    "type": "string",
                    "description": "The search query to look up"
                }
            },
            "required": ["query"]
        }
    }
}
```

# All tools together

```python
tools = [calculator_tool, weather_tool, search_tool]

print(f"Total tools available to LLM: {len(tools)}")
for tool in tools:
    name = tool["function"]["name"]
    desc = tool["function"]["description"]
    print(f"  - {name}: {desc}")
```

## Good vs Bad tool descriptions

BAD  → "description": "Gets weather"

GOOD → "description": "Gets current weather for a city. Use this when user asks about weather, temperature, or climate in any location."

BAD  → "description": "Calculator"

GOOD → "description": "Performs basic math. Use this whenever the user asks to calculate, compute, or solve a math problem."

The description is how the LLM decides WHEN to use the tool.
More specific = better decisions.
