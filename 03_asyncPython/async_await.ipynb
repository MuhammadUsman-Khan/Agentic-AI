# Async/Await Basics

Synchronous code = one thing at a time, in order.
Asynchronous code = start something, don't wait, do other things.

Real world analogy:
- Sync  → you order coffee, stand at counter, wait, get coffee, sit down
- Async → you order coffee, sit down, do other things, get called when ready

# Sync vs Async Comparison

```python
import time
import asyncio

def sync_task(name, duration):
    print(f"{name} started")
    time.sleep(duration)
    print(f"{name} ended")

def run_sync():
    start = time.time()
    sync_task("Task A", 2)
    sync_task("Task B", 2)
    sync_task("Task C", 2)
    print(f"Sync total time: {time.time() - start:.2f}s")

run_sync()
```

```python
async def async_task(name, duration):
    print(f"{name} started")
    await asyncio.sleep(duration)
    print(f"{name} done")
    return f"{name} result"

async def run_async():
    start = time.time()
    results = await asyncio.gather(
        async_task("Task A", 2),
        async_task("Task B", 2),
        async_task("Task C", 2),
    )
    print(f"Async total time: {time.time() - start:.2f}s")
    return results

results = await run_async()
print(results)
```

## What just happened?
- All 3 tasks STARTED at the same time
- Each paused at await asyncio.sleep() without blocking others
- All 3 finished in ~2s instead of ~6s

This is exactly what your agent does when calling
multiple tools or APIs simultaneously.

# Coroutine vs calling it

```python
async def greet(name):
    await asyncio.sleep(1)
    return f"Hello {name}"

coro = greet("Usman")
print(type(coro))

result = await greet("Usman")
print(result)
```
