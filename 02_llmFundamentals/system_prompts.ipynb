# System Prompts & Roles

The chat format has 3 roles:
- system   → instructions to the LLM (personality, rules, context)
- user     → what the human says
- assistant → what the LLM replies

The system prompt is your most powerful tool.
It controls everything — tone, behavior, format, limitations.

# Role Structure

```python
messages = [
    {
        "role": "system",
        "content": "You are a helpful assistant that replies in bullet points only."
    },
    {
        "role": "user",
        "content": "What are the benefits of async programming?"
    }
]

print("Message structure is ready")
print(f"Roles used: {[m['role'] for m in messages]}")
```

## System Prompt Patterns

1. Persona       — "You are an expert Python developer"
2. Format        — "Always reply in JSON"
3. Constraints   — "Never mention competitor products"
4. Context       — "You are helping a beginner learn AI"
5. Chain rules   — "Think step by step before answering"

# Compairing system prompts

```python
base_user_message = "What is recursion?"

prompt_variations = [
    "You are a helpful assistant.",
    "You are a Python tutor for absolute beginners. Use simple language and analogies.",
    "You are a senior software engineer. Be concise and technical.",
    "You are a funny teacher. Explain with humor and real-world examples."
]

for prompt in prompt_variations:
    print(f"System: {prompt[:60]}...")
    print(f"Expected behavior: different tone, depth, and style\n")
```

# Multi-turn conversation structure
This is how a real conversation looks in memory. Each turn adds to the messages list.

```python
conversation = [
    {"role": "system", "content": "You are a concise AI tutor."},
    {"role": "user", "content": "What is a token?"},
    {"role": "assistant", "content": "A token is a chunk of text — roughly 4 characters or 0.75 words. LLMs process tokens, not words."},
    {"role": "user", "content": "Why does it matter?"},
]

print("Conversation turns:", len(conversation))
print("The LLM sees ALL of this on every call - that's the context window")
```
