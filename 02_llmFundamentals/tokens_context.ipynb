# Tokens & Context Window

LLMs don't read words — they read TOKENS.
A token is roughly 4 characters or 0.75 words.

Why does this matter?
- Every API call costs tokens (input + output)
- Every model has a max context window (e.g. 8192 tokens)
- Your agent's memory is limited by this window

## Context Window
The context window is the maximum number of tokens an LLM can
"see" at once - both your input AND its output combined.

Example — llama3-8b-8192 on Groq:
- Max context = 8192 tokens
- If your input = 7000 tokens, output can only be ~1192 tokens
- Old messages get cut off when the window fills up

This is why memory management is a core agentic AI skill.

# Token Counter Helper

```python
def count_tokens(text: str) -> int:
    encoder = tiktoken.get_encoding("cl100k_base")
    return len(encoder.encode(text))

def check_context_limit(messages: list, limit: int = 8192) -> dict:
    total = sum(count_tokens(m["content"]) for m in messages)
    remaining = limit - total 
    return {
        "total_tokens": total,
        "remaining_tokens": remaining,
        "is_safe": remaining > 500
    }

messages = [
    {"role": "system", "content": "You are a helpful AI assistant."},
    {"role": "user", "content": "Explain what Agentic AI means in detail."}
]

status = check_context_limit(messages)
print(f"Total tokens used: {status['total_tokens']}")
print(f"Remaining: {status['remaining_tokens']}")
print(f"Safe to proceed: {status['is_safe']}")
```
