# First LLM Call

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "What is agnetic AI in 2 sentences?"}
    ]
)

print(response.choices[0].message.content)
```

## Breaking down the response object

response.choices[0].message.content  → the actual text reply

response.usage.prompt_tokens          → tokens you sent

response.usage.completion_tokens      → tokens LLM generated

response.usage.total_tokens           → total used

response.model                        → which model responded

# Inspect the full response

```python
print("Reply:", response.choices[0].message.content)
print("---")
print("Prompt tokens:", response.usage.prompt_tokens)
print("Completion tokens:", response.usage.completion_tokens)
print("Total tokens:", response.usage.total_tokens)
print("Model used:", response.model)
```

# Temperature & max_tokens

Temperature controls randomness:
- 0.0 = always same answer (good for agents, facts, JSON output)
- 0.7 = balanced (good for general use)
- 1.0 = very creative/random (good for creative writing)

# Deterministic response

```python
response_det = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
{"role": "user", "content": "Give me a random number between 1 and 10"}
],
    temperature=0.0,
    max_tokens=50
)
```

# Creative response

```python
response_cre = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {"role": "user", "content": "Give me a random number between 1 and 10"}
    ],
    temperature=1.0,
    max_tokens=50
)
```

```python
print("Deterministic", response_det.choices[0].message.content)
print("Creative", response_cre.choices[0].message.content)
```

# Reuseable chat function


```python
def chat(messages: list, temperature: float = 0.7, max_tokens: int = 1000) -> str:
    response = client.chat.completions.create(model="openai/gpt-oss-20b",
     messages=messages,
     temperature=temperature,
    max_tokens=max_tokens
)
    return response.choices[0].message.content
```

```python
reply = chat([
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What is langGraph in one sentences?"}
])

print(reply)
```
